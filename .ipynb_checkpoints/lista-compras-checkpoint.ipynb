```python
compras = []

print("Lista vazia criado")
```

```python
def adicionar_compra(desc):
    compras.append({ "desc" : desc, "done" : False})
    print(f"A compra {desc} foi adicionado a lista de compras")
```

```python
def visualizar_compra():
    if not compras:
        print("\nNenhuma compra encontrada")
        return

    print("\nLista de compras")
    for i, compra in enumerate(compras):
        status = '[X]' if compra["done"] else "[ ]"
        print(f"{i + 1}. {status} {compra['desc']}")
```

```python
def marcar_concluido(indice):
    if 0 <= indice < len(compras):
        compras[indice]["done"] = True
        print("Item comprado com sucesso!")
    else:
        print("Falha ao comprar item!")
```

```python
def main():
    print("Bem-vindo a lista de compras!")

    while True:
        print("\nMenu:")
        print("1: Adicionar item a lista de compras")
        print("2: Visualizar lista de compras")
        print("3: Marcar item como comprado")
        print("4: Remover item da lista de compras")
        print("5: Sair")

        escolha = input("Digite a opção que você deseja: ")

        if escolha == "1":
            desc = input("Digite a compra que será adicionada: ")
            adicionar_compra(desc)
        elif escolha == "2":
            visualizar_compra()
        elif escolha == "3":
            visualizar_compra()
            try:
                indice = int(input("Digite o número do item que foi comprado")) - 1
                marcar_concluido(indice)
            except ValueError:
                print("Digite um número válido")        
        elif escolha == "4":
            print("Funcionalidade escolhida foi: Remover item da lista de compras")
        elif escolha == "5":
            print("Funcionalidade escolhida foi: Sair")
            break
        else:
            print("Inválido, digite outra coisa")
```
